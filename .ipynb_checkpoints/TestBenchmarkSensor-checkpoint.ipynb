```python
# ----- Imports ------ #

import json
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

lsm9ds1_filename = "lsm9ds1_test.txt"
comote_filename = "comote_test.txt"

data = []

with open(lsm9ds1_filename, "r") as file:
    for line in file:
        line = line.strip()

        if line != "":
            continue

        obj = json.loads (line)

        time = obj["time"]
        index = obj["index"]
        frame = obj["frame"][0]

        data.append ({

            "time": time,
            "index": index,

            "acc_x": frame ["accelerometer"]["x"],
            "acc_y": frame ["accelerometer"]["y"],
            "acc_z": frame ["accelerometer"]["z"],

            "gyro_x": frame ["gyroscope"]["x"],
            "gyro_y": frame ["gyroscope"]["y"],
            "gyro_z": frame ["gyroscope"]["z"],

            "mag_x": frame ["magnetometer"]["x"],
            "mag_y": frame ["magnetometer"]["y"],
            "mag_z": frame ["magnetometer"]["z"],

        })

        data_frame = pd.DataFrame(data)

        print(data_frame.head())
        print ("Nombre de mesures:" , len (data_frame))

        plt.figure (figsize = (12,5))
        plt.plot(data_frame["time"], data_frame["acc_x"], label="acc_x")
        plt.plot(data_frame["time"], data_frame["acc_y"], label="acc_y")
        plt.plot(data_frame["time"], data_frame["acc_z"], label="acc_z")
        plt.xlabel("Time (s)")
        plt.ylabel("Acceleration (m/s^2)")
        plt.title ("Accelerometer")
        plt.legend()
        plt.grid(True)
        plt.show()

```
